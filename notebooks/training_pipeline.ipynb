# AquaScan — EfficientNetB0 Training Pipeline
**Person A (ML Engineer) Notebook**

This notebook walks through the full model lifecycle:
1. Environment setup & GPU check
2. Dataset exploration and class distribution
3. Phase 1 training (frozen base)
4. Phase 2 fine-tuning (top-25 layers)
5. Evaluation: accuracy, F1, confusion matrix
6. Grad-CAM visual explainability
7. Export to `model/weights/efficientnetb0_aquascan.h5`

```python
# Cell 1: Environment & GPU Check
import sys, pathlib
print(f'Python: {sys.version}')

import tensorflow as tf
print(f'TensorFlow: {tf.__version__}')
gpus = tf.config.list_physical_devices('GPU')
print(f'GPUs available: {gpus if gpus else "None — using CPU"}')

import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['figure.facecolor'] = '#0e1117'
matplotlib.rcParams['axes.facecolor']   = '#0e1117'
matplotlib.rcParams['text.color']       = '#f0f6fc'
matplotlib.rcParams['axes.labelcolor']  = '#c9d1d9'
matplotlib.rcParams['xtick.color']      = '#8b949e'
matplotlib.rcParams['ytick.color']      = '#8b949e'
```

```python
# Cell 2: Dataset Exploration
import pathlib, collections
from PIL import Image

DATA_DIR = pathlib.Path('../data/dataset')
CLASSES  = ['plastic', 'metal', 'glass', 'cardboard', 'paper', 'trash']

for split in ['train', 'val', 'test']:
    counts = {}
    for cls in CLASSES:
        d = DATA_DIR / split / cls
        counts[cls] = len(list(d.glob('*'))) if d.exists() else 0
    total = sum(counts.values())
    print(f'{split:>5}: {total:>5} total  | ' + '  '.join(f'{c}:{n}' for c,n in counts.items()))

# Visualize class balance
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for i, split in enumerate(['train', 'val', 'test']):
    cnts = [len(list((DATA_DIR/split/c).glob('*'))) if (DATA_DIR/split/c).exists() else 0 for c in CLASSES]
    axes[i].bar(CLASSES, cnts, color='#1f77b4')
    axes[i].set_title(f'{split.capitalize()} split', color='#f0f6fc')
    axes[i].tick_params(axis='x', rotation=30)
plt.suptitle('Class Distribution per Split', color='#f0f6fc', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Cell 3: Sample Grid — 4 random images per class
fig, axes = plt.subplots(len(CLASSES), 4, figsize=(10, 14))
import random
for r, cls in enumerate(CLASSES):
    cls_dir = DATA_DIR / 'train' / cls
    imgs = list(cls_dir.glob('*'))[:4] if cls_dir.exists() else []
    for c in range(4):
        ax = axes[r][c]
        ax.axis('off')
        if c < len(imgs):
            ax.imshow(Image.open(imgs[c]).resize((112,112)))
        if c == 0:
            ax.set_ylabel(cls, color='#f0f6fc', rotation=90, labelpad=10, fontsize=10)
plt.suptitle('Sample Training Images (4 per Class)', color='#f0f6fc', y=1.01)
plt.tight_layout()
plt.show()
```

```python
# Cell 4: Build Data Pipelines
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications.efficientnet import preprocess_input

IMG_SIZE   = (224, 224)
BATCH_SIZE = 32
SEED       = 42

ds_kwargs = dict(
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='categorical',
    class_names=CLASSES,
    seed=SEED,
)

train_ds = keras.utils.image_dataset_from_directory(DATA_DIR/'train', **ds_kwargs)
val_ds   = keras.utils.image_dataset_from_directory(DATA_DIR/'val',   **ds_kwargs)
test_ds  = keras.utils.image_dataset_from_directory(DATA_DIR/'test',  **ds_kwargs)

augment = keras.Sequential([
    layers.RandomFlip('horizontal_and_vertical'),
    layers.RandomRotation(0.15),
    layers.RandomZoom(0.12),
    layers.RandomBrightness(0.15),
    layers.RandomContrast(0.15),
], name='augmentation')

AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.map(lambda x,y: (preprocess_input(tf.cast(augment(x, training=True), tf.float32)), y), num_parallel_calls=AUTOTUNE).cache().shuffle(1000).prefetch(AUTOTUNE)
val_ds   = val_ds.map(lambda x,y: (preprocess_input(tf.cast(x, tf.float32)), y), num_parallel_calls=AUTOTUNE).cache().prefetch(AUTOTUNE)
test_ds  = test_ds.map(lambda x,y: (preprocess_input(tf.cast(x, tf.float32)), y)).prefetch(AUTOTUNE)
print('Pipelines ready')
```

```python
# Cell 5: Build EfficientNetB0 Model
from tensorflow.keras.applications import EfficientNetB0

base = EfficientNetB0(include_top=False, weights='imagenet',
                     input_shape=(*IMG_SIZE, 3), drop_connect_rate=0.2)
base.trainable = False    # Phase 1: frozen

inputs  = keras.Input(shape=(*IMG_SIZE, 3))
x       = base(inputs, training=False)
x       = layers.GlobalAveragePooling2D()(x)
x       = layers.BatchNormalization()(x)
x       = layers.Dropout(0.3)(x)
outputs = layers.Dense(len(CLASSES), activation='softmax', dtype='float32')(x)

model = keras.Model(inputs, outputs, name='AquaScan_EfficientNetB0')
model.summary()
print(f'Total params: {model.count_params():,}')
print(f'Trainable  params: {sum(tf.size(w).numpy() for w in model.trainable_weights):,}')
```

```python
# Cell 6: Phase 1 — Head-only Training
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
import pathlib

WEIGHTS_PATH = pathlib.Path('../model/weights/efficientnetb0_aquascan.h5')
WEIGHTS_PATH.parent.mkdir(parents=True, exist_ok=True)

model.compile(
    optimizer=keras.optimizers.Adam(1e-3),
    loss='categorical_crossentropy',
    metrics=['accuracy'],
)

callbacks_p1 = [
    EarlyStopping(monitor='val_accuracy', patience=4, restore_best_weights=True, verbose=1),
    ModelCheckpoint(str(WEIGHTS_PATH), monitor='val_accuracy', save_best_only=True, verbose=1),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, verbose=1),
]

print('Phase 1: Training classification head ...')
hist1 = model.fit(train_ds, validation_data=val_ds, epochs=10, callbacks=callbacks_p1)
```

```python
# Cell 7: Phase 2 — Fine-tuning Top 25 Layers
base_model = model.layers[1]
base_model.trainable = True
for layer in base_model.layers[:-25]:
    layer.trainable = False

model.compile(
    optimizer=keras.optimizers.Adam(1e-5),
    loss='categorical_crossentropy',
    metrics=['accuracy'],
)

callbacks_p2 = [
    EarlyStopping(monitor='val_accuracy', patience=6, restore_best_weights=True, verbose=1),
    ModelCheckpoint(str(WEIGHTS_PATH), monitor='val_accuracy', save_best_only=True, verbose=1),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, verbose=1),
]

print('Phase 2: Fine-tuning top-25 conv layers ...')
hist2 = model.fit(train_ds, validation_data=val_ds, epochs=15, callbacks=callbacks_p2)
```

```python
# Cell 8: Training Curves
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

p1_acc  = hist1.history['accuracy']
p1_vacc = hist1.history['val_accuracy']
p2_acc  = hist2.history['accuracy']
p2_vacc = hist2.history['val_accuracy']

all_acc  = p1_acc + p2_acc
all_vacc = p1_vacc + p2_vacc
ep = list(range(1, len(all_acc)+1))
boundary = len(p1_acc)

axes[0].plot(ep, all_acc,  label='Train Acc',  color='#58a6ff')
axes[0].plot(ep, all_vacc, label='Val Acc',    color='#3fb950', linestyle='--')
axes[0].axvline(boundary, color='#f0883e', linestyle=':', label='Phase 2 start')
axes[0].set_title('Accuracy', color='#f0f6fc')
axes[0].legend()

p1_loss  = hist1.history['loss']
p2_loss  = hist2.history['loss']
p1_vloss = hist1.history['val_loss']
p2_vloss = hist2.history['val_loss']
all_loss  = p1_loss + p2_loss
all_vloss = p1_vloss + p2_vloss

axes[1].plot(ep, all_loss,  label='Train Loss', color='#58a6ff')
axes[1].plot(ep, all_vloss, label='Val Loss',   color='#3fb950', linestyle='--')
axes[1].axvline(boundary, color='#f0883e', linestyle=':', label='Phase 2 start')
axes[1].set_title('Loss', color='#f0f6fc')
axes[1].legend()

plt.suptitle('EfficientNetB0 Training Curves', color='#f0f6fc')
plt.tight_layout()
plt.savefig('../model/weights/training_curves.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Cell 9: Test Set Evaluation
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

y_true, y_pred = [], []
for batch_x, batch_y in test_ds:
    preds = model.predict(batch_x, verbose=0)
    y_true.extend(tf.argmax(batch_y, axis=1).numpy())
    y_pred.extend(tf.argmax(preds,    axis=1).numpy())

print(classification_report(y_true, y_pred, target_names=CLASSES))

cm = confusion_matrix(y_true, y_pred)
fig, ax = plt.subplots(figsize=(7, 5.5))
fig.patch.set_facecolor('#0e1117')
ax.set_facecolor('#0e1117')
sns.heatmap(cm, annot=True, fmt='d', cmap='mako',
            xticklabels=[c.capitalize() for c in CLASSES],
            yticklabels=[c.capitalize() for c in CLASSES],
            ax=ax, linewidths=0.5, linecolor='#21262d',
            annot_kws={'size':10, 'weight':'bold', 'color':'#ffffff'})
ax.set_title('Test Set Confusion Matrix', color='#f0f6fc', fontsize=12, pad=14, fontweight='bold')
ax.set_xlabel('Predicted', color='#c9d1d9')
ax.set_ylabel('True', color='#c9d1d9')
plt.setp(ax.get_xticklabels(), rotation=30, ha='right', color='#8b949e')
plt.setp(ax.get_yticklabels(), rotation=0,  color='#8b949e')
plt.tight_layout()
plt.savefig('../model/weights/confusion_matrix.png', dpi=150, bbox_inches='tight')
plt.show()
```
